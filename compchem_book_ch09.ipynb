```
def show_3D(mol: psi4.core.Molecule) -> py3Dmol.view:
    """
    Psi4のMoleculeオブジェクトをpy3Dmolで描画する
    Args:
        mol: 描画対象の分子

    Returns:
        py3Dmol.view: py3Dmolの描画オブジェクト

    """
    view = py3Dmol.view(width=400, height=400)
    xyz = mol.save_string_xyz_file()
    view.addModel(xyz, 'xyz')
    view.setStyle({'stick': {}})
    view.setBackgroundColor('#e1e1e1')
    view.zoomTo()

    return view.show()
```

```
def optfreq(mol: psi4.core.Molecule,
            theory: str) -> list[float, psi4.core.Wavefunction]:
    """
    分子の構造最適化と振動数計算を行う

    Args:
        mol (psi4.core.Molecule): 計算対象の分子
        theory: 計算レベル

    Returns:
        float: エネルギー
        psi4.core.Wavefunction: 計算のWavefunction

    """
    _, wfn = psi4.optimize(theory,
                           molecule=mol,
                           return_wfn=True)
    energy, wfn = psi4.frequency(theory,
                                 molecule=mol,
                                 ref_gradient=wfn.gradient(),
                                 return_wfn=True)

    return [energy, wfn]


def get_freqs(wavefunction: psi4.core.Wavefunction) -> np.ndarray:
    """
    Wavefunctionオブジェクトから振動数を取り出す
    Args:
        wavefunction: 対象のWavefunctionオブジェクト

    Returns:
        振動数のarrray

    """
    return wavefunction.frequencies().to_array()
```

### Wavefunctionオブジェクトのvariables

```
def calc_energy(mol: psi4.core.Molecule, theory: str = 'hf/sto-3g') -> psi4.core.Wavefunction:
    """
    エネルギー計算を行い，Wavefunctionオブジェクトを返す
    Args:
        mol: 計算を行う分子
        theory: 計算レベル

    Returns: エネルギー計算で得られるWavefunctionオブジェクト

    """
    _, wfn = psi4.energy(theory, molecule=mol, return_wfn=True)

    return wfn
```

```
methods = ['hf', 'mp2', 'b3lyp']
wfns: list[psi4.core.Wavefunction] = []
```

```
h2o = psi4.geometry('''
0 1
 O                 -0.16182573   -0.32780082    0.00000000
 H                  0.79817427   -0.32780082    0.00000000
 H                 -0.48228031    0.57713501    0.00000000
 ''')
```

```
# エネルギー計算
for method in methods:
    psi4.set_output_file(f'h2o_{method}.log')
    theory = f'{method}/sto-3g'
    wfn = calc_energy(h2o, theory=theory)
    wfns.append(wfn)
```

```
# Wavefunctionオブジェクトのvariablesを出力
for method, wfn in zip(methods, wfns):
    print(f'================   {method}   ================')
    for key in wfn.variables():
        print(key, '\t', wfn.variable(key))
    print()
```

```
def draw_ESP(geom: str = 'geom.xyz',
             cube: str = 'ESP.cube',
             dens: str = 'Dt.cube',
             opacity: float = 0.8,
             cmap: str = 'rwb') -> py3Dmol.view:
    """
    py3DmolでESPマップを描画する
    Args:
        geom: 対象分子のXYZファイル
        cube: 対象分子のcubeファイル
        dens: 対象分子の電子密度ファイル
        opacity: 透過度
        cmap: 描画に用いるcmap（rwb, roygb）

    Returns: py3Dmol.view

    """
    view = py3Dmol.view()
    with open(geom, 'r') as f:
        geom = f.read()
    view.addModel(geom, 'xyz')
    with open(cube, 'r') as f:
        esp = f.read()
    with open(dens, 'r') as f:
        density = f.read()

    view.addVolumetricData(density, 'cube',
                           {'isoval': 0.003,
                            'smoothness': 2,
                            'opacity': opacity,
                            'voldata': esp,
                            'volformat': 'cube',
                            'volscheme': {'gradient': cmap,
                                          'min': -0.05,
                                          'max': 0.05}}
                           )
    view.setStyle({'stick': {}})
    view.setBackgroundColor('#e1e1e1')
    view.zoomTo()

    return view.show()
```
